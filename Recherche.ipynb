```python
def ouverture_fichier(nom):
    with open(nom, 'r', encoding='ISO-8859-1') as fichier:
        contenu = fichier.read()
    return(contenu)
```

```python
contenu = ouverture_fichier("C:/Nouveau dossier (29)/train/24_VO.txt")
```

```python
import os

def lire_documents_dossier(dossier):
    # Initialiser une liste pour stocker les contenus des fichiers
    contenus_fichiers = {}
    
    # Parcourir le répertoire spécifié
    for nom_fichier in os.listdir(dossier):
        # Construire le chemin complet du fichier
        chemin_fichier = os.path.join(dossier, nom_fichier)
        
        # Vérifier si c'est bien un fichier
        if os.path.isfile(chemin_fichier):
            # Ajouter le contenu du fichier dans la liste
            contenus_fichiers[nom_fichier] = {}
            contenus_fichiers[nom_fichier]["content"] = ouverture_fichier(chemin_fichier)
    
    return contenus_fichiers

# Appel de la fonction
FullContenus = lire_documents_dossier('./train')
```

```python
def Get_TF(recherche,contenue):
    weight = 0
    contentarray = recherche.lower().split()
    contenuelowered = contenue.lower().split()
    for word in contentarray:
        weight += contenuelowered.count(word)

    return weight/len(contenuelowered)
```

```python
from cmath import log


def Get_TF_IDF(recherche,docs):
    countTotal = len(docs)
    countIDF = 0
    for docKey in docs:
        print(f"processing {docKey} frequency")
        docs[docKey]["tf"] = Get_TF(recherche,docs[docKey]["content"])
        if docs[docKey]["tf"] != 0:
            countIDF += 1
    IDF = log(countTotal/countIDF).real
    for docKey in docs:
        print(f"processing {docKey} tf-idf")
        docs[docKey]["tf_idf"] = IDF*docs[docKey]["tf"]
    
    return docs
```

```python
results = Get_TF_IDF(recherche2,FullContenus)
```

```python
def trier_par_tf_idf(dictionnaire, n):
    # Trier le dictionnaire par la valeur du champ "tf_idf"
    dictionnaire_trie = sorted(dictionnaire.items(), key=lambda x: x[1]['tf_idf'], reverse=True)
    
    # Prendre les n premières clés (noms des documents)
    noms_documents = [x[0] for x in dictionnaire_trie[:n]]
    
    return noms_documents
```

```python
top10 = trier_par_tf_idf(results, 10)
```
